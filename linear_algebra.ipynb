```python
vector_u = np.array([2,3,2,32,2])
vector_v = np.array([1,2,3,4,5])
```

```python
#calculating dot product of two vectors
def calculate_dot(vector_a, vector_b):
    assert vector_a.shape == vector_b.shape
    n = vector_a.shape[0]

    result = 0.0

    for i in range(n):
        result = result + vector_a[i] * vector_b[i]
    return result
```

```python
calculate_dot(vector_u, vector_v)
```

```python
vector_v.dot(vector_u) # by using builtin numpy function
```
